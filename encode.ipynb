```python
for col in df4.columns:
     if is_numeric_dtype(df4[col]):
          continue
     else:
          uni= df4[col].nunique()
          li= df4[col].unique().tolist()
          if uni==1:
               df4[col]= df4[col].replace(li,[1])
          elif uni==2:
               df4[col]= df4[col].replace(li,[0,1])
          elif uni==3:
               df4[col]= df4[col].replace(li,[0,1,3])
          elif uni==4:
               df4[col]= df4[col].replace(li,[0,1,3,4])
```
